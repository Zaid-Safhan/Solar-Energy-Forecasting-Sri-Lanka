# Data Preparation


```python
import os
import pandas as pd
import numpy as np
```

Set Data Folder Paths


```python
open_meteo_folder = "../Data/Raw/Open_Meteo"
pvgis_folder = "../Data/Raw/PVGIS"

processed_folder = "../Data/Processed"

os.makedirs(
    processed_folder,
    exist_ok=True
)
```

Identify Open-Meteo and PVGIS Files

```python
open_meteo_files = sorted([
    os.path.join(open_meteo_folder, file)
    for file in os.listdir(open_meteo_folder)
    if file.endswith(".csv")
])

pvgis_files = sorted([
    os.path.join(pvgis_folder, file)
    for file in os.listdir(pvgis_folder)
    if file.endswith(".csv")
])

print("Open-Meteo files:", len(open_meteo_files))
print("PVGIS files:", len(pvgis_files))
```

 Load and Combine Open-Meteo Data

```python
open_meteo_list = []

for file in open_meteo_files:
    df = pd.read_csv(file)
    open_meteo_list.append(df)

open_meteo_data = pd.concat(
    open_meteo_list,
    ignore_index=True
)

print("Open-Meteo dataset shape:")
print(open_meteo_data.shape)
```

Load and Combine PVGIS Data

```python
pvgis_list = []

for file in pvgis_files:
    df = pd.read_csv(file)
    pvgis_list.append(df)

pvgis_data = pd.concat(
    pvgis_list,
    ignore_index=True
)

print("PVGIS dataset shape:")
print(pvgis_data.shape)
```

Inspect the Combined Datasets

```python
print("Open-Meteo columns:")
print(open_meteo_data.columns.tolist())

print("\nPVGIS columns:")
print(pvgis_data.columns.tolist())
```

Check District Coverage

```python
open_meteo_district_counts = (
    open_meteo_data["district"]
    .value_counts()
    .sort_index()
)

open_meteo_district_counts
```

```python
pvgis_district_counts = (
    pvgis_data["district"]
    .value_counts()
    .sort_index()
)

pvgis_district_counts
```

Initial Data Quality Check


```python
print("Open-Meteo missing values:")
print(open_meteo_data.isnull().sum())

print("\nPVGIS missing values:")
print(pvgis_data.isnull().sum())
```

```python
print(
    "Open-Meteo duplicated rows:",
    open_meteo_data.duplicated().sum()
)

print(
    "PVGIS duplicated rows:",
    pvgis_data.duplicated().sum()
)
```

Check Data Types


```python
print("Open-Meteo data types:")
print(open_meteo_data.dtypes)

print("\nPVGIS data types:")
print(pvgis_data.dtypes)
```

Prepare Timestamps

```python
open_meteo_data["time_utc"] = pd.to_datetime(
    open_meteo_data["time"],
    utc=True
)

pvgis_data["pvgis_time_utc"] = pd.to_datetime(
    pvgis_data["time"],
    format="%Y%m%d:%H%M",
    utc=True
)
```

```python
print("Open-Meteo:")
print(
    open_meteo_data[
        ["time", "time_utc"]
    ].head()
)

print("\nPVGIS:")
print(
    pvgis_data[
        ["time", "pvgis_time_utc"]
    ].head()
)
```

Align PVGIS Timestamps

```python
pvgis_data["time_utc"] = (
    pvgis_data["pvgis_time_utc"]
    + pd.Timedelta(minutes=30)
)
```

```python
pvgis_data[
    ["time", "pvgis_time_utc", "time_utc"]
].head()
```

Check Timestamp Ranges

```python
print("Open-Meteo:")
print(
    open_meteo_data["time_utc"].min(),
    "to",
    open_meteo_data["time_utc"].max()
)

print("\nPVGIS original:")
print(
    pvgis_data["pvgis_time_utc"].min(),
    "to",
    pvgis_data["pvgis_time_utc"].max()
)

print("\nPVGIS aligned:")
print(
    pvgis_data["time_utc"].min(),
    "to",
    pvgis_data["time_utc"].max()
)
```

Check Timestamp Duplicates

```python
print(
    "Open-Meteo duplicate district/time:",
    open_meteo_data.duplicated(
        subset=["district", "time_utc"]
    ).sum()
)

print(
    "PVGIS duplicate district/time:",
    pvgis_data.duplicated(
        subset=["district", "time_utc"]
    ).sum()
)
```

Check Timestamp Matching

```python
timestamp_check = pd.merge(
    open_meteo_data[
        ["district", "time_utc"]
    ],
    pvgis_data[
        ["district", "time_utc"]
    ],
    on=["district", "time_utc"],
    how="outer",
    indicator=True
)

timestamp_check["_merge"].value_counts()
```

```python
timestamp_check[
    timestamp_check["_merge"] != "both"
].sort_values(
    ["district", "time_utc"]
).head(10)
```

Prepare PVGIS Columns

```python
pvgis_for_merge = pvgis_data[
    [
        "district",
        "time_utc",
        "pvgis_time_utc",
        "P",
        "G(i)",
        "H_sun",
        "T2m",
        "WS10m",
        "Int"
    ]
].copy()
```

```python
pvgis_for_merge = pvgis_for_merge.rename(
    columns={
        "G(i)": "pvgis_gi",
        "H_sun": "pvgis_h_sun",
        "T2m": "pvgis_t2m",
        "WS10m": "pvgis_ws10m",
        "Int": "pvgis_int"
    }
)
```

## Merge Open-Meteo and PVGIS Data

```python
prepared_data = pd.merge(
    open_meteo_data,
    pvgis_for_merge,
    on=["district", "time_utc"],
    how="inner"
)

print("Merged dataset shape:")
print(prepared_data.shape)
```

```python
prepared_data.head()
```

```python
print(
    "Duplicate district/time:",
    prepared_data.duplicated(
        subset=["district", "time_utc"]
    ).sum()
)

print(
    "Districts:",
    prepared_data["district"].nunique()
)
```

Add Sri Lanka Local Time

```python
prepared_data["time_sl"] = (
    prepared_data["time_utc"]
    .dt.tz_convert("Asia/Colombo")
)
```

```python
prepared_data[
    [
        "time_utc",
        "time_sl",
        "district",
        "P"
    ]
].head(10)
```

Check Missing Values After Merge

Check Numerical Ranges

```python
columns_to_check = [
    "temperature_2m",
    "relative_humidity_2m",
    "precipitation",
    "cloud_cover",
    "wind_speed_10m",
    "shortwave_radiation",
    "direct_radiation",
    "diffuse_radiation",
    "direct_normal_irradiance",
    "global_tilted_irradiance",
    "sunshine_duration",
    "is_day",
    "P"
]

prepared_data[columns_to_check].describe().T
```

Check Invalid Values

```python
print(
    "Cloud cover below 0:",
    (prepared_data["cloud_cover"] < 0).sum()
)

print(
    "Cloud cover above 100:",
    (prepared_data["cloud_cover"] > 100).sum()
)

print(
    "Humidity below 0:",
    (prepared_data["relative_humidity_2m"] < 0).sum()
)

print(
    "Humidity above 100:",
    (prepared_data["relative_humidity_2m"] > 100).sum()
)

print(
    "Negative precipitation:",
    (prepared_data["precipitation"] < 0).sum()
)

print(
    "Negative sunshine duration:",
    (prepared_data["sunshine_duration"] < 0).sum()
)

print(
    "Sunshine duration above 3600:",
    (prepared_data["sunshine_duration"] > 3600).sum()
)

print(
    "Negative PV output:",
    (prepared_data["P"] < 0).sum()
)
```

```python
print(
    "is_day values:",
    sorted(
        prepared_data["is_day"]
        .dropna()
        .unique()
    )
)
```

Clean Cloud Cover

```python
prepared_data["cloud_cover"] = (
    prepared_data["cloud_cover"]
    .clip(lower=0, upper=100)
)
```

```python
print(
    "Cloud cover minimum:",
    prepared_data["cloud_cover"].min()
)

print(
    "Cloud cover maximum:",
    prepared_data["cloud_cover"].max()
)
```

Check Missing Precipitation

```python
missing_precipitation = prepared_data[
    prepared_data["precipitation"].isnull()
]

print(
    "Missing precipitation rows:",
    len(missing_precipitation)
)

print(
    "First missing time:",
    missing_precipitation["time_utc"].min()
)

print(
    "Last missing time:",
    missing_precipitation["time_utc"].max()
)

print(
    "Affected districts:",
    missing_precipitation["district"].nunique()
)
```

Check Infinite Values

```python
numeric_data = prepared_data.select_dtypes(
    include=np.number
)

print(
    "Infinite values:",
    np.isinf(numeric_data).sum().sum()
)
```

Check PVGIS Target

```python
print(
    "Minimum P:",
    prepared_data["P"].min()
)

print(
    "Maximum P:",
    prepared_data["P"].max()
)

print(
    "Missing P:",
    prepared_data["P"].isnull().sum()
)

print(
    "Negative P:",
    (prepared_data["P"] < 0).sum()
)

print(
    "Zero P:",
    (prepared_data["P"] == 0).sum()
)
```

Define Target Variable

```python
target_column = "P"

print("Target variable:", target_column)
print("Target unit: W")
print("Reference PV system: 1 kWp")
```

Final Data Check

```python
print("Prepared dataset shape:", prepared_data.shape)
print("Districts:", prepared_data["district"].nunique())

print(
    "Duplicate district/time:",
    prepared_data.duplicated(
        subset=["district", "time_utc"]
    ).sum()
)

print(
    "Missing precipitation:",
    prepared_data["precipitation"].isnull().sum()
)

print(
    "Missing target:",
    prepared_data["P"].isnull().sum()
)

print(
    "Cloud cover range:",
    prepared_data["cloud_cover"].min(),
    "to",
    prepared_data["cloud_cover"].max()
)
```

```python
prepared_file = os.path.join(
    processed_folder,
    "solar_prepared.csv"
)

prepared_data.to_csv(
    prepared_file,
    index=False
)

print("Saved:", prepared_file)
print("Rows:", len(prepared_data))
print("Columns:", len(prepared_data.columns))
```
