```python
import os
import pandas as pd
import requests
```

Adding Locations

```python
districts = [
    "Colombo",
    "Gampaha",
    "Kalutara",
    "Kandy",
    "Matale",
    "Nuwara Eliya",
    "Galle",
    "Matara",
    "Hambantota",
    "Jaffna",
    "Kilinochchi",
    "Mannar",
    "Mullaitivu",
    "Vavuniya",
    "Batticaloa",
    "Ampara",
    "Trincomalee",
    "Kurunegala",
    "Puttalam",
    "Anuradhapura",
    "Polonnaruwa",
    "Badulla",
    "Monaragala",
    "Ratnapura",
    "Kegalle"
]

location_data = pd.DataFrame(
    districts,
    columns=["district"]
)

location_data
```

### Get coordinates for the district locations

```python
url = "https://geocoding-api.open-meteo.com/v1/search"

search_names = {
    "Mullaitivu": "Mullaittivu"
}

latitudes = []
longitudes = []
elevations = []

for district in location_data["district"]:
    search_name = search_names.get(district, district)

    params = {
        "name": search_name,
        "count": 10,
        "language": "en",
        "format": "json",
        "countryCode": "LK"
    }

    data = requests.get(url, params=params).json()
    results = data.get("results", [])

    match = None

    for place in results:
        if place.get("name", "").lower() == search_name.lower():
            match = place
            break

    if match:
        latitudes.append(match["latitude"])
        longitudes.append(match["longitude"])
        elevations.append(match.get("elevation"))

        print(
            district,
            match["latitude"],
            match["longitude"],
            match.get("elevation")
        )
    else:
        latitudes.append(None)
        longitudes.append(None)
        elevations.append(None)

        print(district, "Not found")
```

```python
location_data["latitude"] = latitudes
location_data["longitude"] = longitudes
location_data["elevation"] = elevations

location_data
```

Checking Null and Unique Values

```python
print("Number of districts:", len(location_data))
print("Unique districts:", location_data["district"].nunique())

print("\nMissing values:")
print(location_data.isnull().sum())

print("\nDuplicate coordinates:")
print(location_data.duplicated(["latitude", "longitude"]).sum())
```

```python
os.makedirs(
    "../Data/Reference",
    exist_ok=True
)

location_data.to_csv(
    "../Data/Reference/location_master.csv",
    index=False
)
```
